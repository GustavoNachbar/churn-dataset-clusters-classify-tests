```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

variaveis_numericas = [
    'PontuacaoCredito',
    'Idade',
    'TempoRelacionamento',
    'Saldo',
    'NumeroProdutos',
    'SalarioEstimado'
]

df_treino_cancelamento = df_treino[df_treino["Saiu"] == 1][variaveis_numericas]


# X = df_treino[df_treino["Saiu"] == 1][variaveis_numericas]

# Padronização
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_treino_cancelamento)

# K-Means com 3 clusters
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df_treino_cancelamento['Cluster'] = kmeans.fit_predict(X_scaled)

df_treino_cancelamento['Distancia_Centroide'] = kmeans.transform(X_scaled).min(axis=1)
```

```python
variaveis_numericas = [
    'PontuacaoCredito',
    'Idade',
    'TempoRelacionamento',
    'Saldo',
    'NumeroProdutos',
    'SalarioEstimado'
]

resumo_clusters = df_treino_cancelamento.groupby('Cluster').agg(
    Quantidade=('Cluster', 'size'),
    **{
        variavel: (variavel, 'mean')
        for variavel in variaveis_numericas
    }
)

resumo_clusters['Representatividade'] = (
    resumo_clusters['Quantidade'] / len(df_treino_cancelamento) * 100
)

resumo_clusters
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.utils import resample
from sklearn.metrics import adjusted_rand_score
import numpy as np

# --- Métricas internas (usando o cluster já gerado com k=3) ---
labels = df_treino_cancelamento['Cluster']

silhouette = silhouette_score(X_scaled, labels)
db_index = davies_bouldin_score(X_scaled, labels)
ch_index = calinski_harabasz_score(X_scaled, labels)

print(f"Silhouette Score       : {silhouette:.4f}")
print(f"Davies-Bouldin Index    : {db_index:.4f}")
print(f"Calinski-Harabasz Index : {ch_index:.2f}")


# --- Estabilidade (bootstrap + ARI) ---
def calcular_estabilidade(X, k, labels_referencia, n_bootstrap=30, sample_frac=0.8, random_state=42):
    rng = np.random.default_rng(random_state)
    n_samples = X.shape[0]
    ari_scores = []

    for i in range(n_bootstrap):
        seed = int(rng.integers(0, 1_000_000))
        idx_sample = resample(
            np.arange(n_samples),
            replace=True,
            n_samples=int(n_samples * sample_frac),
            random_state=seed,
        )
        idx_sample = np.unique(idx_sample)

        X_sample = X[idx_sample]

        kmeans_boot = KMeans(n_clusters=k, random_state=seed, n_init=10)
        labels_boot = kmeans_boot.fit_predict(X_sample)

        labels_ref_sample = np.asarray(labels_referencia)[idx_sample]
        ari = adjusted_rand_score(labels_ref_sample, labels_boot)
        ari_scores.append(ari)

    return np.mean(ari_scores), np.std(ari_scores)


estab_media, estab_std = calcular_estabilidade(X_scaled, k=3, labels_referencia=labels)
print(f"Estabilidade (ARI médio): {estab_media:.4f} ± {estab_std:.4f}")
```

```python
# Seleciona as mesmas variáveis utilizadas no treinamento
X_treino = df_treino[variaveis_numericas]

# Padroniza utilizando o scaler JÁ ajustado nos dados de cancelamento
X_treino_scaled = scaler.transform(X_treino)

# Identifica o cluster mais próximo de cada cliente
df_treino['Cluster'] = kmeans.predict(X_treino_scaled)

# Calcula a distância até o centroide do cluster atribuído
distancias = kmeans.transform(X_treino_scaled)

df_treino['Distancia_Centroide'] = [
    distancias[i, cluster]
    for i, cluster in enumerate(df_treino['Cluster'])
]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# 1. VARIÁVEIS DO MODELO
# ============================================================

variaveis_numericas = [
    'PontuacaoCredito',
    'Idade',
    'TempoRelacionamento',
    'Saldo',
    'NumeroProdutos',
    'SalarioEstimado',
    'Cluster',
    'Distancia_Centroide'
]

variaveis_categoricas = [
    'Geografia',
    'Genero'
]

variaveis_modelo = variaveis_numericas + variaveis_categoricas


# ============================================================
# 2. DEFINIÇÃO DE X E Y
# ============================================================

X = df_treino[variaveis_modelo]
y = df_treino['Saiu']


# ============================================================
# 3. DIVISÃO TREINO E TESTE
# ============================================================

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# 4. TRATAMENTO DAS VARIÁVEIS CATEGÓRICAS
# ============================================================

preprocessador = ColumnTransformer(
    transformers=[
        (
            'categoricas',
            OneHotEncoder(handle_unknown='ignore'),
            variaveis_categoricas
        )
    ],
    remainder='passthrough'
)


# ============================================================
# 5. CRIAÇÃO DA ÁRVORE DE DECISÃO
# ============================================================

arvore = DecisionTreeClassifier(
    random_state=42,
    class_weight='balanced'
)


# ============================================================
# 6. PIPELINE
# ============================================================

modelo = Pipeline(
    steps=[
        ('preprocessador', preprocessador),
        ('arvore', arvore)
    ]
)


# ============================================================
# 7. TREINAMENTO
# ============================================================

modelo.fit(X_treino, y_treino)


# ============================================================
# 8. PREVISÃO
# ============================================================

y_pred = modelo.predict(X_teste)

y_prob = modelo.predict_proba(X_teste)[:, 1]


# ============================================================
# 9. AVALIAÇÃO
# ============================================================

print(f'Acurácia: {accuracy_score(y_teste, y_pred):.2%}')

print('\nRelatório de Classificação:')
print(classification_report(y_teste, y_pred))

print('\nMatriz de Confusão:')
print(confusion_matrix(y_teste, y_pred))


# ============================================================
# 10. PROBABILIDADE DE CANCELAMENTO PARA TODA A BASE
# ============================================================

df_treino['Probabilidade_Cancelamento'] = (
    modelo.predict_proba(
        df_treino[variaveis_modelo]
    )[:, 1]
)


# ============================================================
# 11. PREVISÃO DE CANCELAMENTO PARA TODA A BASE
# ============================================================

df_treino['Previsao_Cancelamento'] = (
    modelo.predict(
        df_treino[variaveis_modelo]
    )
)
```

# TREIANDO ALGORITIMO DE ARVORE DE DECISÃO COM GRID SEARCH

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
import pandas as pd

variaveis_numericas = [
    'PontuacaoCredito', 'Idade', 'TempoRelacionamento',
    'Saldo', 'NumeroProdutos', 'SalarioEstimado'
]

y = df_treino['Saiu']

# Split único, reaproveitado em todos os k's (garante que a comparação seja justa)
idx_train, idx_test = train_test_split(
    df_treino.index, test_size=0.2, random_state=42, stratify=y
)

param_grid = {
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 5],
    'criterion': ['gini', 'entropy']
}

resultados_arvore = []
melhores_modelos = {}

for k in range(2, 8):
    features = variaveis_numericas + [f'Cluster_k{k}', f'Distancia_Centroide_k{k}']

    X_train = df_treino.loc[idx_train, features]
    X_test = df_treino.loc[idx_test, features]
    y_train = y.loc[idx_train]
    y_test = y.loc[idx_test]

    grid_search = GridSearchCV(
        estimator=DecisionTreeClassifier(random_state=42),
        param_grid=param_grid,
        cv=5,
        scoring='f1',   # f1 é mais informativo que accuracy quando as classes são desbalanceadas
        n_jobs=-1
    )
    grid_search.fit(X_train, y_train)

    melhor_modelo = grid_search.best_estimator_
    melhores_modelos[k] = melhor_modelo

    y_pred = melhor_modelo.predict(X_test)
    y_proba = melhor_modelo.predict_proba(X_test)[:, 1]

    resultados_arvore.append({
        'k': k,
        'melhores_parametros': grid_search.best_params_,
        'accuracy': accuracy_score(y_test, y_pred),
        'precision': precision_score(y_test, y_pred),
        'recall': recall_score(y_test, y_pred),
        'f1': f1_score(y_test, y_pred),
        'roc_auc': roc_auc_score(y_test, y_proba),
    })

df_resultados_arvore = pd.DataFrame(resultados_arvore).set_index('k')
df_resultados_arvore
```

```python
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
import pandas as pd

param_grid_svm = {
    'svm__C': [0.1, 1, 10],
    'svm__kernel': ['linear', 'rbf'],
    'svm__gamma': ['scale', 'auto']
}
# 3 x 2 x 2 = 12 combinações x 5 folds = 60 modelos por k
# (o SVM é mais lento por amostra que os algoritmos anteriores, mas o grid é bem menor)

resultados_svm = []
melhores_modelos_svm = {}

for k in range(2, 8):
    features_numericas = variaveis_numericas + [f'Distancia_Centroide_k{k}']
    cluster_dummies = pd.get_dummies(df_treino[f'Cluster_k{k}'], prefix=f'cluster_k{k}')

    X = pd.concat([df_treino[features_numericas], cluster_dummies], axis=1)

    X_train = X.loc[idx_train]
    X_test = X.loc[idx_test]
    y_train = y.loc[idx_train]
    y_test = y.loc[idx_test]

    pipeline_svm = Pipeline([
        ('scaler', StandardScaler()),
        ('svm', SVC(probability=True, random_state=42))
    ])

    grid_search_svm = GridSearchCV(
        estimator=pipeline_svm,
        param_grid=param_grid_svm,
        cv=5,
        scoring='f1',
        n_jobs=-1
    )
    grid_search_svm.fit(X_train, y_train)

    melhor_modelo = grid_search_svm.best_estimator_
    melhores_modelos_svm[k] = melhor_modelo

    y_pred = melhor_modelo.predict(X_test)
    y_proba = melhor_modelo.predict_proba(X_test)[:, 1]

    resultados_svm.append({
        'k': k,
        'melhores_parametros': grid_search_svm.best_params_,
        'accuracy': accuracy_score(y_test, y_pred),
        'precision': precision_score(y_test, y_pred),
        'recall': recall_score(y_test, y_pred),
        'f1': f1_score(y_test, y_pred),
        'roc_auc': roc_auc_score(y_test, y_proba),
    })

df_resultados_svm = pd.DataFrame(resultados_svm).set_index('k')
df_resultados_svm
```
